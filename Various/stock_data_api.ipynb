```python
#API to get the indian market stock price
import yfinance as yf
import logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s',filename='stock_prices.log')
data = yf.download(["RELIANCE.NS","SBIN.NS","PNB.NS"], period="20y", interval="1d")
print(data)
logging.info("Stock prices fetched successfully")
```

```python
tickers = ["RELIANCE.NS","SBIN.NS","PNB.NS"]
ticker_object = yf.Tickers(tickers)
logging.info("History-{}".format(ticker_object.history(period="20y", interval="1d")))
ticker_object.tickers['RELIANCE.NS'].info
ticker_object.tickers['SBIN.NS'].info
ticker_object.tickers['PNB.NS'].info    
print(ticker_object.tickers['RELIANCE.NS'].get_actions())
ticker_object.tickers['SBIN.NS'].actions
ticker_object.tickers['PNB.NS'].actions
ticker_object.tickers['RELIANCE.NS'].analyst_price_targets
ticker_object.tickers['SBIN.NS'].analyst_price_targets
ticker_object.tickers['PNB.NS'].analyst_price_targets
print(ticker_object.tickers['RELIANCE.NS'].get_analyst_price_targets())
```
