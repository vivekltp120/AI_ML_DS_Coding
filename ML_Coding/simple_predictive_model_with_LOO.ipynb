```python
import numpy as np

# Example data array
data = np.array([10, 20, 30, 40, 50])

# Placeholder to store predictions
predictions = []
errors = []

print("Performing Leave-One-Out Cross Validation...\n")

# LOO Cross-Validation
for i in range(len(data)):
    # Leave one value out
    train_data = np.delete(data, i)  # All elements except the i-th
    test_value = data[i]            # The left-out element

    # Calculate the mean of the training data
    predicted_value = np.mean(train_data)

    # Store the prediction and error
    predictions.append(predicted_value)
    error = abs(predicted_value - test_value)
    errors.append(error)

    print(f"Iteration {i + 1}:")
    print(f"  Train Data: {train_data}")
    print(f"  Test Value: {test_value}")
    print(f"  Predicted Value: {predicted_value:.2f}")
    print(f"  Absolute Error: {error:.2f}\n")

# Final Results
average_error = np.mean(errors)
print(f"Predictions: {np.array(predictions).mean()}")
print(f"Mean Absolute Error (MAE): {average_error:.2f}")
```
