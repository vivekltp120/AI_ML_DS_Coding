```python
import pandas as pd
from sklearn.model_selection import train_test_split 
df = pd.read_csv("creditcard.csv")
attributes=len(df.columns)
input_data=df.iloc[:,0:attributes-2]
target_data=df.iloc[:,-1]
print(input_data)
print(target_data)
from collections import Counter
print(Counter(input_data))

X_train,X_test,y_train,y_test=train_test_split(input_data,target_data,test_size=.30,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from imblearn.ensemble import BalancedBaggingClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score,classification_report
from sklearn import tree
from matplotlib import pyplot as plt
model=RandomForestClassifier(random_state=42)
# model_imbalance=GridSearchCV
model=model.fit(X_train,y_train)
y_test_predict=model.predict(X_test)
y_train_predict=model.predict(X_train)

print("test Accuracy:\n",accuracy_score(y_true=y_test,y_pred=y_test_predict))
print("training Accuracy:\n",accuracy_score(y_true=y_train,y_pred=y_train_predict))
```

```python
#SMOTE for imbalance 
from imblearn.over_sampling import SMOTE
from collections import Counter
# Display class distribution before SMOTE
print("Class distribution before SMOTE:", Counter(y_train))
smote=SMOTE(random_state=42,sampling_strategy='auto')
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)
print("Class distribution after SMOTE:", Counter(y_train_resampled))
model=RandomForestClassifier(random_state=42)
# model_imbalance=GridSearchCV
model=model.fit(X_train_resampled, y_train_resampled)
y_test_predict=model.predict(X_test)
y_train_predict=model.predict(X_train)

print("Test Accuracy-",classification_report(y_true=y_test,y_pred=y_test_predict))
print("Test Accuracy-",classification_report(y_true=y_train,y_pred=y_train_predict))
```
