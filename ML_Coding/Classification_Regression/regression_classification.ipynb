```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import load_diabetes
diabetes_data=load_diabetes()
```

```python
from sklearn.model_selection import train_test_split 
from sklearn.metrics import accuracy_score,mean_squared_error,mean_absolute_error
# print(diabetes_data)
input_data=diabetes_data.data
target_data=diabetes_data.target
X_train,X_test,y_train,y_test=train_test_split(input_data,target_data,test_size=0.33,random_state=42)

parameters={"max_depth":[1,2,3,4,5,6,7,8,9],"splitter":['random','best'],"criterion":["squared_error", "friedman_mse", "absolute_error", "poisson"]}
            
diabetes_regressor=DecisionTreeRegressor()

from sklearn.model_selection import GridSearchCV
model=GridSearchCV(diabetes_regressor,param_grid=parameters,cv=10)
model.fit(X_train,y_train)
print(model.best_params_)
y_predict_test=model.predict(X_test)
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_pred=y_predict_test,y_true=y_test)
mae=mean_absolute_error(y_pred=y_predict_test,y_true=y_test)
print(mse)
print(mae)
```
