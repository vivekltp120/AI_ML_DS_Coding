```python
from matplotlib import pyplot as plt
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report
from sklearn.datasets import load_wine
from sklearn import tree
data=load_wine()
X_data=data.data
Y_data=data.target
print(X_data)
X_train,X_test,y_train,y_test=train_test_split(X_data,Y_data,test_size=0.2,stratify=Y_data,random_state=42)
parameters={"objective":"multi:softprob","num_classes":3,"max_depth":5}
model=xgb.XGBClassifier(parameters)
model.fit(X_train,y_train)
y_predict=model.predict(X_test)
accuracy=accuracy_score(y_pred=y_predict,y_true=y_test)
report=classification_report(y_pred=y_predict,y_true=y_test)
print(f'accuracy-{accuracy} \n report-{report}')




```
