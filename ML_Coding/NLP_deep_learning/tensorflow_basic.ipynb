```python
#model training with keras
from tensorflow import keras
from tensorflow.keras import datasets
from tensorflow.keras.datasets import mnist
from tensorflow.keras import layers

(x_train,y_train),(x_test,y_test)=mnist.load_data()
print(x_train.shape)
print(x_test.shape)

#reshape the column
x_train=x_train.reshape(-1,28*28).astype('float32')/255.0
x_test=x_test.reshape(-1,28*28).astype('float32')/255.0
```

```python
#Sequential API
model= keras.models.Sequential(
    [ keras.Input(shape=(784,)),
    layers.Dense(512,name='First Layer',activation='tanh'),
     layers.Dense(256,name='Second Layer',activation='tanh'),
     layers.Dense(10,name='Output Layer')]
)

model.compile(loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),optimizer=keras.optimizers.Adam(learning_rate=0.001),metrics=['accuracy'])
model.fit(x_train,y_train,batch_size=32,epochs=7,verbose=2)
model.evaluate(x_test,y_test,batch_size=32,verbose=2)
model.summary()
```

```python
#Functional API (Flexible)
a0=keras.Input(shape=(784,))
a1=layers.Dense(512,activation='relu')(a0)
a2=layers.Dense(256,activation='relu')(a1)
output=layers.Dense(10,name='output',activation='softmax')(a2)
model=keras.Model(inputs=a0,outputs=output)


model.compile(loss=keras.losses.SparseCategoricalCrossentropy(from_logits=False),optimizer=keras.optimizers.Adam(learning_rate=0.001),metrics=['accuracy'])
model.fit(x_train,y_train,batch_size=32,epochs=7,verbose=2)
model.summary()
model.evaluate(x_test,y_test,batch_size=32,verbose=2)
```

```python
test=x_test[:3]
print(test)
result=model.predict(test)
result
```
