```python
import requests
import bs4
url="https://www.analyticsvidhya.com/blog/2023/01/top-100-data-science-interview-questions/"
res=requests.get(url)
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(res.content, 'html5lib') # If this line causes an error, run 'pip install html5lib' or install html5lib
#print(soup)
```

```python
content=[]
for row in soup.findAll('div',attrs={"class":"accordion-item mb-4 border-0 rounded-3"}):
    content_dict={}
    # print(row.prettify())
    content_dict['About']=row.h5.text.strip()
    content_dict['Reference']=row.a['href']
    # content_dict['Div']=row.div.div['class']   
    content.append(content_dict)
print(content)
    


    
```

```python
import pandas as pd
df=pd.DataFrame(content)
```

```python
df.to_csv("analytic_vidya_ml.csv")
df.to_excel("analytic_vidya.xlsx",sheet_name="machine_learning")
```
