```python
import numpy as np
a=np.array([[1,2,3,4],[2,3,4,5]])
print(np.sort((np.log(a))))
print(np.exp(a))
print(np.square(a))
print(np.sqrt(a))
print(np.abs(a))
print(np.shape(a))

b=np.array([4,5,6,7])
print(np.shape(b))
b.reshape(4,1)
print(b.shape)

c=np.dot(a,b)
print(c)
print(a.size)

d=np.dot(b.T,a.T)
print(d)
```

```python
#sigma funtion implemenation
# %matplotlib inline
A=1/(1+np.exp(-b))
print(A)
import matplotlib.pyplot as plt

plt.plot(b,A)
plt.xlabel("z")
plt.ylabel("sigmoid z")
plt.show()
```

```python
import time
a=np.random.rand(1000000)
b=np.random.random(1000000)
tic=time.time()
c=np.dot(a,b)
toc=time.time()
print("Vectorization version - ",toc-tic)
c=0
tic=time.time()
for i in range(1000000):
   c+=a[i]*b[i]
toc=time.time()
print("Non Vectorization version - ",toc-tic)

    

```
