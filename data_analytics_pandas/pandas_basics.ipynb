```python
# get a metrix and change substitue o with column mean.
import numpy as np
import pandas as pd
random_matrix=np.random.randint(0,10,size=(3,3))
print(random_matrix)
matrix=pd.DataFrame(random_matrix)
print(matrix.sum(axis=1))
for column in matrix.columns:
    value=matrix[column].mean()
    matrix[column]=matrix[column].replace(0,value)
matrix.var(axis=0)    
```
