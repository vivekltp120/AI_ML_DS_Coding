```python
#For each batsman
#Sum of Runs for all batsman
stats_player['batsman_runs'].sum().sort_values(ascending=False)
#Top 10 Six Hitter Batsman
cond4=delivery['batsman_runs']==4
cond6=delivery['batsman_runs']==6

six_stats=delivery[cond6].groupby('batter')
six_stats['batsman_runs'].count().sort_values(ascending=False).head(10).to_dict()
figure=plt.figure(figsize=(8,6))
sns.barplot(x=list(six_stats['batsman_runs'].count().sort_values(ascending=False).head(10).index), y=list(six_stats['batsman_runs'].count().sort_values(ascending=False).head(10).values))
plt.title('Top 10 Six Hitter Batsmen')
plt.xlabel('Batsman')
plt.ylabel('Number of Sixes')
plt.xticks(rotation=60)
plt.show()
```

```python
#last five over highest strick rate batsman
mask1=delivery['over']>15
stats_last_over=delivery[mask1]
stats_last_over.shape
bat_final=stats_last_over.groupby('batter')
#Srick Rate for each batsman
last_over_balls=bat_final['ball'].count()
last_over_runs=bat_final['batsman_runs'].sum()
final_df=pd.concat([last_over_balls,last_over_runs],axis=1)
final_df['strick_rate']=last_over_runs/last_over_balls*100

#batsman who played more than 100 ball in final over NOT Required For Now
last_over_mask_100=bat_final['ball'].count()>100
last_over_batsman=last_over_balls[last_over_mask_100]
last_over_batsman_list=last_over_batsman.index.tolist()

# batsman strick rate  for the batsman played more then 100 balls
final_df[final_df.ball>100].sort_values('strick_rate',ascending=False).reset_index().head(10)
```

```python
#merge on the datasets
match_stats=match.merge(delivery,left_on='id',right_on='match_id')
match_stats.reset_index().head(1)
match_stats.columns
```

```python
#match stats for the batsman
match_stats_batsman=match_stats.groupby('batter')
match_stats_batsman['batsman_runs'].sum().sort_values(ascending=False).head(10).to_dict() 

plt.figure(figsize=(8,6))
sns.barplot(x=list(match_stats_batsman['batsman_runs'].sum().sort_values(ascending=False).head(10).index), y=list(match_stats_batsman['batsman_runs'].sum().sort_values(ascending=False).head(10).values))
plt.title('Top 10 Batsmen by Runs')
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.xticks(rotation=60)
plt.show()
```
